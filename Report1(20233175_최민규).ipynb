```python
n = int(input("정방행렬의 차수를 입력하세요: "))

matrix = []

for i in range(n):
    row = list(map(float, input(f"{i + 1}행: ").split()))
    matrix.append(row)
```

```python
def determinant(matrix):
    n = len(matrix)

    # 1x1 행렬
    if n == 1:
        return matrix[0][0]

    # 2x2 행렬
    if n == 2:
        return matrix[0][0] * matrix[1][1] - matrix[0][1] * matrix[1][0]

    det = 0
    for j in range(n):
        # 부분행렬 만들기
        sub_matrix = [
            [matrix[i][k] for k in range(n) if k != j]
            for i in range(1, n)
        ]

        # 여인수 전개
        cofactor = (-1) ** j * determinant(sub_matrix)

        det += matrix[0][j] * cofactor

    return det
```

```python
def inverse_by_determinant(matrix):
    n = len(matrix)

    det = determinant(matrix)

    # 역행렬 존재 여부 확인
    if abs(det) == 0:
        raise ValueError("역행렬이 존재하지 않는 행렬입니다.")

    # 여인수 행렬
    cofactor_matrix = []

    for i in range(n):
        row = []

        for j in range(n):
            sub_matrix = [
                [matrix[r][c] for c in range(n) if c != j]
                for r in range(n) if r != i
            ]

            cofactor = (-1) ** (i + j) * determinant(sub_matrix)
            row.append(cofactor)

        cofactor_matrix.append(row)

    # 여인수 행렬의 전치 = 수반행렬
    adjoint = [
        [cofactor_matrix[j][i] for j in range(n)]
        for i in range(n)
    ]

    # A^-1 = adj(A) / det(A)
    inverse = [
        [adjoint[i][j] / det for j in range(n)]
        for i in range(n)
    ]

    return inverse
```

```python
def inverse_by_gauss_jordan(matrix):
    n = len(matrix)

    # [A | I] 만들기
    augmented = []

    for i in range(n):
        row = matrix[i][:]

        for j in range(n):
            if i == j:
                row.append(1.0)
            else:
                row.append(0.0)

        augmented.append(row)

    # 가우스-조던 소거
    for i in range(n):

        # 현재 열에서 절댓값이 가장 큰 행 찾기
        pivot_row = i

        for r in range(i + 1, n):
            if abs(augmented[r][i]) > abs(augmented[pivot_row][i]):
                pivot_row = r

        # 피벗이 0이면 역행렬 없음
        if abs(augmented[pivot_row][i]) < 1e-10:
            raise ValueError("역행렬이 존재하지 않는 행렬입니다.")

        # 행 교환
        augmented[i], augmented[pivot_row] = \
            augmented[pivot_row], augmented[i]

        # 피벗을 1로 만들기
        pivot = augmented[i][i]

        for j in range(2 * n):
            augmented[i][j] /= pivot

        # 다른 행의 해당 열을 0으로 만들기
        for r in range(n):
            if r != i:
                factor = augmented[r][i]

                for j in range(2 * n):
                    augmented[r][j] -= factor * augmented[i][j]

    # 오른쪽 부분만 추출
    inverse = []

    for i in range(n):
        inverse.append([round(x, 10) for x in augmented[i][n:]])

    return inverse
```

```python
inverse1 = inverse_by_determinant(matrix)
inverse2 = inverse_by_gauss_jordan(matrix)
print("행렬식을 이용한 역행렬:")
print(inverse1)

print("가우스-조던 소거법을 이용한 역행렬:")
print(inverse2)

if inverse1 == inverse2:
    print("두 방식의 결과는 동일합니다.")
else:
    print("두 방식의 결과는 다릅니다.")
```
